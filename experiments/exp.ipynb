```python
from scipy.optimize import linear_sum_assignment

wiseLayoutInit = {
    0: (1,1), 1: (1,2), 2: (1,3),
    3: (2,1), 4: (2,2), 5: (2,3),
    6: (3,1), 7: (3,2), 8: (3,3),
}

wiseLayoutTarget = {
    5: (1,1), 6: (1,2), 2: (1,3),
    3: (2,1), 0: (2,2), 7: (2,3),
    1: (3,1), 8: (3,2), 4: (3,3),
}

rows = 3
cols = 3

def pprint(_Tinit):

    for r in range(3):
        print(r+1, [_Tinit[(r+1,c+1)] for c in range(3)])

    print('\n')
            

Tinit = {}
for idx in wiseLayoutInit.keys():
    Tinit[wiseLayoutInit[idx]]=wiseLayoutTarget[idx]


targRowByinitCol = [ [ False for _ in range(cols)] for _ in range(rows)]

for r in range(rows):
    ir = r+1
    lhnodes = []
    rhnodes = []
    for c in range(cols):
        ic = c+1
        if (ir, ic) in Tinit:
            tr, tc = Tinit[ir, ic]
            lhnodes.append((tr, tc))
        rhnodes.append(ic)

    cost_matrix = [[targRowByinitCol[tr-1][ic-1] for ic in rhnodes] for (tr,_) in lhnodes]
    row_ind, col_ind = linear_sum_assignment(cost_matrix)
    print(row_ind)
    
```
